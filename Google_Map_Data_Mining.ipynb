```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from bs4 import BeautifulSoup
import pandas as pd
from time import sleep

# Set up Selenium WebDriver
driver = webdriver.Chrome()  # Replace with the path to your WebDriver
driver.get("https://www.google.com/maps")
sleep(5)  # Allow the page to load

# Define the extensive list of cities
cities = [
    "Talwandi", "Tanda", "Tando Adam", "Tando Allahyar", "Tando Bagho", "Tando Gulam Ali", "Tando Jam", 
    "Tando Muhammad Khan", "Tank", "Tappi", "Tarbela Dam", "Tarnol", "Taxila", "Temar Garah", "Terbela Ghazi", 
    "Thana Bola Khan", "Thanil Fatoohi", "Thari Mirwah", "Tharo Shah", "Thatta", "Thul", "Timergera", "Toba Tek Singh", 
    "Topi", "Tulamba", "Turbat", "Ubaouro", "Uch Sharif", "Ugoki", "Umarkot", "Umarzai", "Upper Dir", "Usta Mohammad", 
    "Uthal", "Vehari", "Waggan", "Wah Cantt", "Waisa", "Warah", "Wasu", "Winder", "Yazman Mandi", "Zafar Wal", "Zahir Pir", 
    "Zhob", "Zia Abad"
]


# Initialize lists to store data
all_data = []

def search_city(city):
    """
    Searches for a specific city in Google Maps.
    """
    try:
        map_searchbar = driver.find_element(By.CLASS_NAME, 'searchboxinput')
        map_searchbar.clear()  # Clear previous input
        map_searchbar.send_keys(f'Mobile Retailers {city} pakistan')
        sleep(2)
        google_map_search_button = driver.find_element(By.CLASS_NAME, 'mL3xi')
        google_map_search_button.click()
        sleep(5)  # Wait for search results to load
    except Exception as e:
        print(f"Error searching for {city}: {e}")

def scroll_and_extract(city):
    """
    Scrolls through the page and extracts data for the current city.
    """
    try:
        scrolling_cube = driver.find_element(By.XPATH, '//*[@id="QA0Szd"]/div/div/div[1]/div[2]/div/div[1]/div/div/div[1]/div[1]')
        for _ in range(20):  # Adjust the range based on the number of listings
            driver.execute_script("arguments[0].scrollTop += 1000", scrolling_cube)
            sleep(2)  # Adjust based on loading speed
        extract_data(city)
    except Exception as e:
        print(f"Error during scrolling for {city}: {e}")

def extract_data(city):
    """
    Extracts data (names, addresses, and phone numbers) for the current city.
    """
    page_source = BeautifulSoup(driver.page_source, 'html.parser')
    results = page_source.find_all('div', class_='bfdHYd Ppzolf OFBs3e')  # Update class if necessary

    for result in results:
        try:
            # Extract Name
            name = result.find('div', class_='NrDZNb')
            name = name.get_text() if name else 'Null'
        except Exception as e:
            print(f"Error extracting name for {city}: {e}")
            name = 'Null'

        try:
            # Extract Address
            address_elements = result.find_all('div', class_='W4Efsd')
            address = address_elements[2].get_text() if len(address_elements) >= 3 else 'Null'
        except Exception as e:
            print(f"Error extracting address for {city}: {e}")
            address = 'Null'

        try:
            # Extract Number
            number_info = result.find('span', class_='UsdlK')
            number = number_info.get_text() if number_info else 'No number found'
        except Exception as e:
            print(f"Error extracting number for {city}: {e}")
            number = 'No number found'

        # Append data with city name
        all_data.append({
            "City": city,
            "Name": name,
            "Address": address,
            "Number": number
        })

# Iterate over the cities
for city in cities:
    print(f"Extracting data for {city}...")
    search_city(city)
    scroll_and_extract(city)

# Create a DataFrame and save the results
df = pd.DataFrame(all_data)
df.to_excel("Google_Maps_City_Data_2.xlsx", index=False)

print("Data extraction complete. Saved to Google_Maps_City_Data_2.xlsx.")

# Close the browser
driver.quit()
```
